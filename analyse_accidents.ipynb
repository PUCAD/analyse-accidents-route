```python
import pandas as pd

chemin = r"C:\Users\miste\Downloads\analyse-accidents-route\data\usagers-2024.csv"
usagers = pd.read_csv(chemin, sep=";", low_memory=False)

print("Dimensions :", usagers.shape)
print(usagers.columns.tolist())
print(usagers.isna().sum())
```

```python
usagers["age"] = 2024 - usagers["an_nais"]

print(usagers["grav"].value_counts().sort_index())
print(usagers["catu"].value_counts().sort_index())
print(usagers["age"].describe())
```

```python
usagers["grave"] = usagers["grav"].isin([2, 3])

print(usagers.groupby("catu")["grave"].mean().round(3))

usagers["tranche_age"] = pd.cut(
    usagers["age"],
    bins=[0, 17, 24, 39, 59, 120],
    labels=["0-17", "18-24", "25-39", "40-59", "60+"]
)
print(usagers.groupby("tranche_age", observed=True)["grave"].mean().round(3))
```

```python
usagers["tranche_age"] = pd.cut(
    usagers["age"],
    bins=[0, 18, 25, 40, 60, 121],
    labels=["0-17", "18-24", "25-39", "40-59", "60+"],
    right=False
)
print(usagers.groupby("tranche_age", observed=True)["grave"].mean().round(3))
```

```python
from scipy.stats import chi2_contingency

tableau = pd.crosstab(usagers["catu"], usagers["grave"])
chi2, p, dof, attendus = chi2_contingency(tableau)
print("p-value :", p)
```
